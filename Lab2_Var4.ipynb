# Практическая работа №2: Группировка, обработка пропусков и объединение данных
**Вариант 4**

```python
import pandas as pd
import numpy as np

# Таблица 1: Продажи
sales = pd.DataFrame({
    'OrderID': [101, 102, 103, 104, 105, 106, 107, 108, 109, 110],
    'Product': ['Laptop', 'Phone', 'Tablet', 'Laptop', 'Phone',
                'Laptop', 'Tablet', 'Phone', 'Laptop', 'Tablet'],
    'Quantity': [1, 2, 1, 3, 1, 2, 1, 4, 1, 2],
    'Price': [50000, 30000, 20000, 50000, np.nan,
              50000, 20000, 30000, 50000, np.nan],
    'Region': ['Moscow', 'SPb', 'Kazan', 'Moscow', 'SPb',
               'Kazan', np.nan, 'SPb', 'Moscow', 'Kazan']
})

# Таблица 2: Информация о товарах
products = pd.DataFrame({
    'Product': ['Laptop', 'Phone', 'Tablet', 'Monitor'],
    'Category': ['Electronics', 'Electronics', 'Electronics', 'Accessories'],
    'Weight_kg': [2.5, 0.2, 0.5, 3.0]
})

print("Таблица продаж:")
print(sales)
print("\nТаблица товаров:")
print(products)
```

## Задание 1. Найти сумму `Price` для каждого региона (`.groupby()`)

```python
print(sales.groupby('Region')['Price'].sum())
```

## Задание 2. Найти количество уникальных продуктов в каждом регионе

```python
unique_products = sales.groupby('Region')['Product'].nunique()
print(unique_products)
```

## Задание 3. Заполнить пропуски в `Price` значением 25000

```python
print('Пропуски в Price до заполнения:', sales['Price'].isnull().sum())

sales['Price'] = sales['Price'].fillna(25000)

print('Пропуски в Price после заполнения:', sales['Price'].isnull().sum())

print("\n")
print(sales)
```

## Задание 4. Преобразовать столбец `Product` в категориальный тип

```python
print('Тип до преобразования:', sales['Product'].dtype)

sales['Product'] = sales['Product'].astype('category')

print('Тип после преобразования:', sales['Product'].dtype)
print('Категории:', list(sales['Product'].cat.categories))

#Все типы
print("\n")
print(sales.dtypes)
```

## Задание 5. Объединить `sales` и `products`, сохранив все записи из `sales` (left join)

```python
merged = pd.merge(sales, products, on='Product', how='left')
print(merged)
```
