# Практическая работа №1: Основы работы с данными в Pandas
**Вариант 4**

```python
import pandas as pd
import numpy as np

data = {
    'ID': range(1, 11),
    'Name': ['Alice', 'Bob', 'Charlie', 'David', 'Eva',
             'Frank', 'Grace', 'Helen', 'Ivan', 'Julia'],
    'Age': [23, 35, 29, 40, 28, 45, 32, 27, 38, 31],
    'City': ['Moscow', 'SPb', 'Moscow', 'Kazan', 'SPb',
             'Moscow', 'SPb', 'Kazan', 'Moscow', 'SPb'],
    'Salary': [50000, 70000, 60000, 80000, 55000,
               90000, 65000, 58000, 75000, 62000]
}

df = pd.DataFrame(data)
print(df)
```

## Задание 1. Вывести типы данных всех столбцов (`.dtypes`)

```python
print(df.dtypes)
```

## Задание 2. Выбрать строку с индексом 5 (используя `.loc[]`)

```python
print(df.loc[5])
```

## Задание 3. Выбрать строки, где Age между 28 и 35 (включительно)

```python
print(df[df['Age'].between(28, 35)])
```

## Задание 4. Найти сумму всех зарплат

```python
print('Сумма всех зарплат:', df['Salary'].sum())
```

## Задание 5. Создать столбец `Name_Length` — длину имени (функция `len`)

```python
df['Name_Length'] = df['Name'].map(len)
print(df)
```
